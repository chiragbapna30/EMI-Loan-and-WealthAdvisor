# EMI and Loan Advisor Agent

## Tools Used
* `compute_emi(amount, rate_annual, months)`: Calculates monthly installment payments (EMI), total repayable amount, and total interest accrued.
* `compare_options(options, monthly_income)`: Compares multiple loan choices side-by-side, applies the 40% EMI-to-income affordability threshold, and identifies the option with the lowest total interest.

## Memory Implementation
The agent utilizes persistent session memory (`self.memory`) to track context across turns:
* **User Context:** Remembers the user's stated monthly income across turns so the user does not need to re-enter it.
* **Option History:** Stores all computed EMI choices during the conversation session to perform instant side-by-side comparative analyses when requested.

## Honest Failure & Resolution
* **Failure:** When comparing a shorter high-EMI tenure with a longer low-EMI tenure, the agent initially recommended the lower EMI option strictly based on monthly affordability, ignoring that the user would pay double the overall interest over time.
* **Handling:** Added explicit total-interest comparison logic into `compare_options()` so the agent explicitly flags interest differences alongside monthly affordability.
